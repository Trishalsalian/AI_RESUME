```
import subprocess, time, re

!pkill -f streamlit
!pkill -f cloudflared
!wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O cloudflared
!chmod +x cloudflared

subprocess.Popen(["streamlit", "run", "app.py", "--server.port", "8501", "--server.headless", "true",
                  "--server.enableCORS", "false", "--server.enableXsrfProtection", "false"],
                 stdout=open("streamlit_log.txt", "w"), stderr=subprocess.STDOUT)
time.sleep(6)
subprocess.Popen(["./cloudflared", "tunnel", "--url", "http://localhost:8501", "--no-autoupdate"],
                 stdout=open("cf_log.txt", "w"), stderr=subprocess.STDOUT)

url = None
for _ in range(30):
    time.sleep(2)
    m = re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", open("cf_log.txt").read())
    if m:
        url = m.group(0); break
print("Open this link:", url or "not found - run !cat cf_log.txt")
```

```
import subprocess, time, re, urllib.request

!pkill -f streamlit
!pkill -f cloudflared
!cp "app (1) (1).py" app.py   # skip this line if app.py is already the new version
!grep -c "Dataset Evaluation" app.py

subprocess.Popen(["python", "-m", "streamlit", "run", "app.py", "--server.port", "8501", "--server.headless", "true",
                  "--server.enableCORS", "false", "--server.enableXsrfProtection", "false"],
                 stdout=open("streamlit_log.txt", "w"), stderr=subprocess.STDOUT)

for _ in range(60):                      # wait up to ~2 minutes for the app
    try:
        urllib.request.urlopen("http://localhost:8501", timeout=2); print("app is up"); break
    except Exception:
        time.sleep(2)
else:
    print("app did not start"); print(open("streamlit_log.txt").read()[-1500:])

subprocess.Popen(["./cloudflared", "tunnel", "--url", "http://localhost:8501", "--no-autoupdate"],
                 stdout=open("cf_log.txt", "w"), stderr=subprocess.STDOUT)
for _ in range(30):
    time.sleep(2)
    m = re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", open("cf_log.txt").read())
    if m: print("Open this link:", m.group(0)); break
```
